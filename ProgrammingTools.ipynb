## Programación funcional

Paradigma que trata la computación como la evaluación de funciones puras, evitando estados mutables y efectos secundarios. Se centra en la composición y el uso de funciones de orden superior para expresar la lógica de forma declarativa.

- Enfoque: funciones puras e inmutabilidad.  
- Conceptos clave: funciones de orden superior, composición, recursión y expresiones declarativas.  
- Beneficios: código más predecible, fácil de razonar y probar; mejor tolerancia a la concurrencia.

### Funciones Anonimas: Lambda
Una función lambda es una función anónima y compacta definida en una sola expresión, creada para ser usada despues

- Sintaxis (Python): `lambda argumentos: expresión`  
- Uso: crear pequeñas funciones puntuales para `map`, `filter`, `sorted`, o callbacks sin usar `def`.  

```python
f = lambda x, y, z: x + y + z
f(2, 3, 4)  #Crear una funcion lambda
```

```python
def func(x, y, z):
    return x + y + z
func(2, 3, 4) #Crear una funcion normal
```

```python
f = (lambda a="fee", b="fie", c="foe": a + b + c)
f("wee") # Esto hace que a="wee", b="fie", c="foe" y suma los tres valores
```

### Mapeado de funciones sobre iterables: map

Map ejecuta una función especifica para cada item en un iterable y es enviado a la funcion como un parametro

```python
def inc(x): return x + 10
list(map(inc, counters))  #map hace lo mismo pero con una funcion normal
```

```python
counters = [1,2,3,4]
updated = []
for x in counters:
    updated.append(x + 10)
updated
```

```python
lista = list(map((lambda x: x + 10), counters))
print(lista)
```

```python
pow(2,4) # Eleva 2 a la potencia de 4

list(map(pow, [2,3,4], [3,4,5]))  # Eleva cada elemento de la primera lista a la potencia del elemento correspondiente en la segunda lista
```

```python
list(map(inc, [1,2,3,4])) # Sumar 10 a cada elemento de la lista usando la funcion inc
```

```python
list(range(-5, 5)) # Genera una lista de numeros desde -5 hasta 4

list(filter((lambda x: x > 0), range(-5, 5))) # Filtra los numeros mayores que 0 de la lista generada por range
```

### Combinar items en iterables: reduce


```python

from functools import reduce
reduce((lambda x, y: x + y), [1, 2, 3, 4]) # Suma todos los elementos de la lista usando reduce
  
```
