## Comprensión de lista y Programación funcional



```python
ord("a")
```

```python
res = []
for x in "spam":
    res.append(ord(x))
res   # Convierte cada caracter en su valor ASCII y lo agrega a la lista res
```

```python
res = list(map(ord, 'spam'))
res  # Convierte cada caracter en su valor ASCII usando map
```

```python
res = [ord(x) for x in 'spam'] # Convierte cada caracter en su valor ASCII usando una comprehension
res
```

```python
[x ** 2 for x in range(10)] # Genera una lista de los cuadrados de los numeros del 0 al 9

```

```python
list(map((lambda x: x ** 2), range(10))) # Genera una lista de los cuadrados de los numeros del 0 al 9 usando map y lambda
```

```python
[x for x in range(5) if x % 2 == 0] # Genera una lista de numeros pares del 0 al 4 usando una comprehension si el resto es 0
list(filter((lambda x: x % 2 == 0), range(5)))
```

```python
[x ** 2 for x in range(10) if x % 2 == 0] # Genera una lista de los cuadrados de los numeros pares del 0 al 9 usando una comprehension con condicion
```

```python
list( map((lambda x: x**2), filter((lambda x: x % 2 == 0), range(10))) )
# Genera una lista de los cuadrados de los numeros pares del 0 al 9 usando map y filter
#map hace la transformacion (cuadrado) y filter filtra los numeros pares
```

```python
res = [x + y for x in [0, 1, 2] for y in [100, 200, 300]]
res # Genera una lista con la suma de cada elemento de la primera lista con cada elemento de la segunda lista
```

```python
[x + y for x in 'spam' for y in 'SPAM']
# Genera una lista con la concatenacion de cada caracter de la primera cadena con cada caracter de la segunda cadena
```

```python
[x + y for x in 'spam' if x in 'sm' for y in 'SPAM' if y in ('P', 'A')]
# Genera una lista con la concatenacion de cada caracter de la primera cadena que este en 'sm' con cada caracter de la segunda cadena que este en ('P', 'A')
```

```python
[(x, y) for x in range(5) if x % 2 == 0 for y in range(5) if y % 2 == 1]
# Genera una lista de tuplas con pares (x, y) donde x es un numero par del 0 al 4 e y es un numero impar del 0 al 4
```

### Example: List Comprehensions and Matrixes


```python
M = [[1, 2, 3],
	 [4, 5, 6],
	 [7, 8, 9]]

N = [[2, 2, 2],
     [3, 3, 3],
	 [4, 4, 4]]

M[1]

M[1][2]
```

```python
[row[1] for row in M] # Printea la segunda columna de cada fila de la matriz M
```

```python
[M[row][1] for row in range(len(M))] # Printea la segunda columna de cada fila de la matriz M usando indices
```

```python
[ M[i][i] for i in range( len(M) ) ] # Printea la diagonal principal de la matriz M
```

```python
[ M[i][len(M)-1-i] for i in range( len(M) ) ] # Printea la diagonal secundaria de la matriz M
```

```python
[col + 10 for row in M for col in row] # Suma 10 a cada elemento de la matriz M y lo devuelve en una lista plana
```

```python
[ [col + 10 for col in row] for row in M] # Suma 10 a cada elemento de la matriz M y lo devuelve en una matriz del mismo tamaño
```

```python
[ M[row][col] * N[row][col] for row in range(3) for col in range(3)]
# Multiplica elemento a elemento las matrices M y N y lo devuelve en una lista plana
```

```python
[[M[row][col] * N[row][col] for col in range(3)] for row in range(3)]
# Multiplica elemento a elemento las matrices M y N y lo devuelve en una matriz del mismo tamaño
```

```python
[[col1 * col2 for (col1, col2) in zip(row1, row2)] for (row1, row2) in zip(M, N)]
# Multiplica elemento a elemento las matrices M y N usando zip y lo devuelve en una matriz del mismo tamaño
```
