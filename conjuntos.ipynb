## SETS

Tipo que permite almacenar y acceder a varios elementos, no hay duplicados, desordenados, inmutables

```python
set([1,2,3,4]) # Crea un conjunto de una lista
```

```python
{1, 2, 3, 4} # Crea un conjunto directamente
```

```python
set("spam") # Crea un conjunto de una cadena de caracteres de forma desordenada
```

```python
S = {"s", "p", "a", "m"}
S  
```

```python
S.add("alot") # Añade un elemento al conjunto
S
```

```python
S1 = {1, 2, 3}
S1 & {1, 3}  # Intersección de conjuntos
```

```python
{3, 6} | S1 # Unión de conjuntos
```

```python
S1 - {1, 3, 4}  # Diferencia de conjuntos
```

```python
S1 > {1,2} # Comprueba si un conjunto es un superconjunto de otro
```

```python
S1 - {1, 2, 3, 4} # Diferencia simétrica de conjuntos
```

```python
type({}) # Tipo de dato conjunto
```

```python
S = set()
S.add(1.23)
S # Muestra el conjunto tras añadir un elemento
```

## Vistas diccionario y sets


```python
D = dict(a=1, b=2, c=3)         # D = {'b': 2, 'c': 3, 'a': 1}
K = D.keys()
V = D.values()
del D['b']
D
```

```python
K, V # Muestra las vistas de claves y valores tras eliminar un par clave-valor
```

```python
K | {'x': 4}            # La vista `keys` es como un `set` ç
V | {3, 4}              # La vista `values` no es como un `set`

 D
```
