## Iterations and Comprehensions

Realizar una operación sobre cada item de una secuencia es una comprehension

```python
L = [1, 2, 3, 4, 5]
for i in range(len(L)): # este range hace que i tome los valores 0, 1, 2, 3, 4
    L[i] += 10 # Agrega 10 al elemento en la posición i de la lista
L
```

```python
L = [x + 10 for x in L] # Agrega 10 a cada elemento x en la lista
L
```

```python
res = [] # lo mismo que lo de arriba pero con un for tradicional
for x in L:
    res.append(x + 10)
res
```

```python
L = [1, 7, 8, 9]
M = [x for x in L if x < 5] # Filtra los elementos de L que son menores que 5 y los guarda en M
M
```

```python
[x + y for x in "abc" for y in "lmn"] # Combina cada letra de "abc" con cada letra de "lmn"
```

```python
res = [] # lo mismo que lo de arriba pero con un for tradicional
for x in 'abc':
	for y in 'lmn':
		res.append(x + y)
res
```

```python
T = (1, 2, 3, 4, 5)
L = [ x + 20 for x in T ] # Suma 20 a cada elemento x en la tupla T y crea una lista
L
```

```python
L = [ x for x in range(0,9) ] # Crea una lista con los números del 0 al 8
L
```

```python
x = [1, 2, 3]
y = [4, 5, 6]
zipped = zip(x, y) # Combina las listas x e y en pares de tuplas
list(zipped)
```

```python
L = [ x for x in range(0,4) ]
M = [ x for x in range(5,10) ]
[elemento for elemento in zip(L,M) ] # Combina las listas L y M en pares de tuplas
```

```python
list(zip(['a', 'b', 'c'], [1, 2, 3])) # Combina dos listas en pares de tuplas
```

```python
D = dict(zip(['a', 'b', 'c'], [1, 2, 3]))	
D # Crea un diccionario a partir de dos listas, una para las claves y otra para los valores
```

```python
D = {k: v for (k, v) in zip(['a', 'b', 'c'], [1, 2, 3])}
D # Crea un diccionario usando una comprensión de diccionarios y zip
```
