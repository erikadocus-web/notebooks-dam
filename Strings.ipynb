# *FUNCIONES STRINGS*

str() cambia un numero a un string

```python
number1 = 12
number1ToString = str(number1)
number1ToString
```

.split() divide el string en arrays apartir del parámetro

```python
text2 = "Hola_Me_Llamo"
texto2Cambiado = text2.split("_")
texto2Cambiado
```

.join() une strings añadiendo lo que ponga antes del punto a lo que está entre parentesis

```python
text3 = "Hola", "Que", "Tal"
text3Cambiado = "_".join(text3)
text3Cambiado
```

replace() cambia el primer parametro por el segundo

```python
text1 = "Hola_Buenos_Días"
texto1Cambiado = text1.replace("_", " ")
texto1Cambiado
```

### Longitud y tipo
`len()` devuelve la longitud y `type()` muestra el tipo de dato.

```python
s = 'Hola Mundo'
len(s)
type(s)
```

### Mayúsculas y minúsculas
Funciones: `upper()`, `lower()`, `capitalize()`, `title()`, `swapcase()` y `casefold()` (más agresiva para comparaciones).

```python
t = 'python es GENIAL'
t.upper()
t.lower()
t.capitalize()
t.title()
t.swapcase()
t.casefold()
```

### Eliminación de espacios y caracteres alrededor (strip)
`strip()` quita espacios al inicio y final; `lstrip()` `rstrip()` para cada lado.

```python
raw = '   hola   '
raw.strip()
raw.lstrip()
raw.rstrip()
```

### Buscar y contar
`find()` devuelve la primera posición (o -1), `index()` lanza excepción si no existe; `count()` cuenta ocurrencias.

```python
texto = 'abracadabra'
texto.find('bra')
texto.count('a')
texto.rfind('a')
# texto.index('z')  # lanzaría ValueError si no lo encuentra
```

### Inicio / final y particiones
`startswith()` y `endswith()` son útiles para comprobaciones; `partition()` divide por el primer separador y `rpartition()` por el último.

```python
frase = 'hola-mundo-ejemplo'
frase.startswith('hola')
frase.endswith('ejemplo')
frase.partition('-')
frase.rpartition('-')
```

### Separar (split) y unir (join)
`split()` ya está en el notebook; aquí un ejemplo con `splitlines()` y `join()` para reconstruir.

```python
multiline = 'uno
dos
tres
'
multiline.splitlines()
'_'.join(['uno','dos','tres'])
```

### Slicing (rebanado) — extracción por índices
Explica `s[start:end:step]` y ejemplos con paso negativo para invertir.

```python
s = 'abcdefgh'
s[0:4]  # primeros 4
s[4:]   # desde el índice 4 hasta el final
s[::2]  # cada 2 caracteres
s[::-1] # invertir
```

### Comprobaciones de contenido (isalpha, isdigit, isalnum, isspace)
Útiles para validar cadenas simples.

```python
'abc'.isalpha()
'123'.isdigit()
'abc123'.isalnum()
'   
'.isspace()
```

### Alineación y relleno
`zfill()`, `center()`, `ljust()`, `rjust()` para dar formatos simples de salida.

```python
'7'.zfill(3)
'x'.center(7, '-')
'x'.ljust(5, '.')
'x'.rjust(5, '.')
```

### Formateo - `format` y f-strings
F-strings son la forma moderna y legible de insertar variables.

```python
nombre = 'Ana'
f'Hola {nombre}!'
'Hola {}'.format(nombre)
```
